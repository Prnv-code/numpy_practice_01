### NUMPY INDEXING AND SLICING OF VECTORS

```python
arr = np.arange(11,21)
arr
```

```python
# indexing
arr[9]
```

```python
# slicing
arr[1:5]
```

```python
arr[:4]
```

```python
arr[:]
```

```python
arr[4:]
```

```python
# to reverse string
reverse = arr[::-1]
```

### NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1, 31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[3]
```

```python
arr[0,0]
```

```python
arr[0,4]
```

```python
arr[5,4]
```

```python
# slicing
slice = arr[0:2,1:3]
slice
```

```python
slice1 = arr[3: , 3: ]
slice1
```

```python
arr[:,2]
```

### BOOLEAN INDEXING

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
