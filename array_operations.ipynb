```python
a1 ** a2
```

## Advance Array Manipulation

### Stacking Array

```python
# for stacking we use vectors, not matrix
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
```

```python
# vertical stack
np.vstack((a, b))
```

```python
# horizontal stack
np.hstack((a,b))
```

```python
# coloum st
np.column_stack((a,b))
```

### Splitting Array

```python
c = np.arange(16).reshape(4,4)
c
```

```python
# horizontal split
np.hsplit(c,2)
```

```python
# vertical splito
np.vsplit(c,2)
```
